```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.decomposition import PCA
import numpy as np

import matplotlib.pyplot as plt
plt.rcParams.update({'font.size': 16})

import seaborn as sns

import matplotlib
%matplotlib inline

import glob
```

```python
allinfo = pd.read_csv('../data/tornados_all.csv')
```

```python
months = np.array(allinfo['MONTH'])
months = [int(i) for i in months]
```

```python
spring = allinfo[(allinfo['MONTH'] == 3) | (allinfo['MONTH'] == 4) | (allinfo['MONTH'] == 5)] 
summer = allinfo[(allinfo['MONTH'] == 6) | (allinfo['MONTH'] == 7) | (allinfo['MONTH'] == 8)]
autumn = allinfo[(allinfo['MONTH'] == 9) | (allinfo['MONTH'] == 10) | (allinfo['MONTH'] == 11)]
winter = allinfo[(allinfo['MONTH'] == 12) | (allinfo['MONTH'] == 1) | (allinfo['MONTH'] == 2)]
```

```python
np.array(winter['TOR_F_SCALE'] >= 2), np.array(winter['YEAR'] == 2000)
```

```python
## https://repository.library.noaa.gov/view/noaa/69284 
```

```python
Nspring = [] # March, April, May
Nsummer = [] # June, July, August
Nautumn = [] # September, October, November
Nwinter = [] # December, January, February
df_all = glob.glob('../data/per_year/*_tornados.csv')

df_all = np.sort(df_all)
years = []

median_month = []
std_month = []

EF0 = []
EF1 = []
EF2 = []
EF3 = []
EF4 = []
EF5 = []

for n in range(0,len(df_all)):
    print(df_all[n])
    df = pd.read_csv(df_all[n])
    year = df['YEAR'][0]
    years.append(year)
    median_month.append(np.mean(df['MONTH']))
    std_month.append(np.std(df['MONTH']))
    
    EF0.append(len(df[df['TOR_F_SCALE'] == 0]))
    EF1.append(len(df[df['TOR_F_SCALE'] == 1]))
    EF2.append(len(df[df['TOR_F_SCALE'] == 2]))
    EF3.append(len(df[df['TOR_F_SCALE'] == 3]))
    EF4.append(len(df[df['TOR_F_SCALE'] == 4]))
    EF5.append(len(df[df['TOR_F_SCALE'] == 5]))
    
    Nspring.append(len( spring[(spring['YEAR'] == year) * (spring['TOR_F_SCALE'] >= 1)] ) )
    Nsummer.append(len( summer[(summer['YEAR'] == year) * (summer['TOR_F_SCALE'] >= 1)] ) )
    Nautumn.append(len( autumn[(autumn['YEAR'] == year) * (autumn['TOR_F_SCALE'] >= 1)] ) )
    Nwinter.append(len( winter[(winter['YEAR'] == year) * (winter['TOR_F_SCALE'] >= 1)] ) )
```

```python
plt.figure()
plt.scatter(years, median_month)
```

```python
plt.figure()
plt.plot(years, Nspring, label='spring')
#plt.figure()
plt.plot(years, Nsummer, label='summer')
#plt.figure()
plt.plot(years, Nautumn, label='autumn')
#plt.figure()
plt.plot(years, Nwinter, label='winter')
plt.xlabel('Year')
plt.legend()
plt.ylabel('Number of Events')
```

```python
plt.figure()
plt.plot(years, EF0, label='EF0')
#plt.figure()
plt.plot(years, EF1, label='EF1')
#plt.figure()
plt.plot(years, EF2, label='EF2')
#plt.figure()
plt.plot(years, EF3, label='EF3')
plt.plot(years, EF4, label='EF4')
plt.plot(years, EF5, label='EF5')
plt.legend()
plt.xlabel('Year')
plt.ylabel('Number of Events')
```

```python
df_all = glob.glob('../data/per_year/*_tornados.csv')

df_all = np.sort(df_all)

cmap = plt.colormaps['winter']

years = []
month_peak = []
bins = np.arange(1,13)

plt.figure()
for n in range(0,len(df_all)):
    print(df_all[n])
    df = pd.read_csv(df_all[n])
    counts, bins = np.histogram(df['MONTH'], bins=bins)
    year = int(df['YEAR'][0])
    color_val = (2025-year)/len(df_all)
    plt.plot(bins[1:], counts, color=cmap(color_val), label=str(year))
    years.append(year)
    month_peak.append(np.argsort(counts)[-1] + 1)
    
plt.legend(fontsize=12, loc='upper right', bbox_to_anchor=(1.3, 1.1))
```
