```python
df_all = glob.glob('../data/per_year/*_tornados.csv')

df_all = np.sort(df_all)

cmap = plt.colormaps['winter']

years = []
month_peak = []
bins = np.arange(1,13)

plt.figure()
for n in range(0,len(df_all)):
    print(df_all[n])
    df = pd.read_csv(df_all[n])
    counts, bins = np.histogram(df['MONTH'], bins=bins)
    year = int(df['YEAR'][0])
    color_val = (2025-year)/len(df_all)
    plt.plot(bins[1:], counts, color=cmap(color_val), label=str(year))
    years.append(year)
    month_peak.append(np.argsort(counts)[-1] + 1)
    
plt.legend(fontsize=12, loc='upper right', bbox_to_anchor=(1.3, 1.1))
```
